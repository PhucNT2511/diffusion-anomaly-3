```python
import torch
import torch.nn.functional as F

# Giả sử batch_size = 2
B = 2
C, H, W = 2, 3, 3  # Số kênh, chiều cao và chiều rộng của ảnh

# Tạo một batch ảnh ngẫu nhiên với requires_grad=True để theo dõi gradient
images = torch.randn(B, C, H, W, requires_grad=True)

# Giả sử chúng ta có một mô hình đơn giản, ở đây là một lớp Conv2d
conv_layer = torch.nn.Conv2d(in_channels=C, out_channels=8, kernel_size=3, stride=1, padding=1)

# Tính toán đầu ra của mô hình
outputs = conv_layer(images)  # Kích thước: (B, 8, 256, 256)

# Tính toán một giá trị mất mát giả định, ví dụ: tổng bình phương của outputs
loss = torch.sum(outputs ** 2)

# Tính gradient của loss đối với images
gradients = torch.autograd.grad(loss, images)[0]

# Kiểm tra kích thước của gradients
print("Kích thước của gradients:", gradients.shape)
```
